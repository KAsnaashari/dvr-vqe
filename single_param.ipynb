```python
def build_circuit(gates, n_qubits=5):
    from qiskit import QuantumCircuit
    from qiskit.circuit import Parameter

    out = QuantumCircuit(5)
    theta = Parameter(f'$\\theta$')
    for g in gates:
        if g[0] == 'cnot':
            out.cnot(*g[1])
            
        if g[0] == 'cz':
            out.cz(*g[1])

        elif g[0] == 'ry':
            out.ry(g[2] * theta, g[1])

        elif g[0] == 'h':
            out.h(g[1])

        elif g[0] == 'y':
            out.y(g[1])

        elif g[0] == 'x':
            out.x(g[1])

        elif g[0] == 'rz':
            out.rz(theta, g[1])
    return out

gates = [['ry', 0, 2], ['cnot', [3, 0]]]
print(build_circuit(gates))
```

```python
from qiskit import QuantumCircuit, transpile
from qiskit.circuit import Parameter
%autoreload

n_qubits = 5
n_select = 3
init_ansatzes = []
for i in range(n_qubits):
    ansatz = QuantumCircuit(n_qubits)
    theta = Parameter(f'$\\theta$')
    ansatz.ry(theta, i)
    init_ansatzes.append(ansatz)

init_best_qubit = None
init_best_energy = None
for i in range(n_qubits):
    print(init_ansatzes[i])
    converge_cnts, converge_vals, best_params, best_energies = dvr_vqe.opt_vqe(h_dvr_pauli, init_ansatzes[i], vqe_options)
    print(best_params)
    if (init_best_energy is None) or (np.min(best_energies) < init_best_energy):
        init_best_qubit = i
        init_best_energy = np.min(best_energies)
    print('-----------------------')
print(init_best_qubit)
print(init_best_energy)

layers = 5
rands = 3
touched_qubits = [set([init_best_qubit]) for i in range(n_select)]
ansatz0s = [init_ansatzes[init_best_qubit]]
layer_energies = [init_best_energy]
for l in range(layers):
    print(f'Layer: {l}')
    ansatzes = []
    gate_list = []
    gate_names = []
    for i in range(n_qubits):
        for a, ansatz0 in enumerate(ansatz0s):  
            for j in touched_qubits[a]:
                if i != j:
                    ansatz = ansatz0.copy()
                    ansatz.cnot(j, i)
                    ansatzes.append(ansatz)
                    gate_list.append([j, i])
                    gate_names.append([a, 'cnot'])
    
    for i in range(n_qubits):
        for a, ansatz0 in enumerate(ansatz0s):
            ansatz = ansatz0.copy()
            ansatz.h(i)
            ansatzes.append(ansatz)
            gate_list.append(i)
            gate_names.append([a, 'h'])
    
    # for i in range(n_qubits):
    #     for a, ansatz0 in enumerate(ansatz0s):
    #         ansatz = ansatz0.copy()
    #         ansatz.x(i)
    #         ansatzes.append(ansatz)
    #         gate_list.append(i)
    #         gate_names.append([a, 'x'])

    for i in range(n_qubits):
        theta = Parameter(f'$\\theta$')
        for a, ansatz0 in enumerate(ansatz0s):
            for rand in range(rands):
                ansatz = ansatz0.copy()
                ansatz.assign_parameters({ansatz.parameters[0]: theta}, inplace=True)
                x = np.random.uniform(low=-2, high=2)
                ansatz.ry(x * theta, i)
                ansatzes.append(ansatz)
                gate_list.append(i)
                gate_names.append([a, 'ry', x])

    print('----------------')

    best_gates = []
    best_energy = []
    for i in range(len(ansatzes)):
        print(ansatzes[i])
        converge_cnts, converge_vals, best_params, best_energies = dvr_vqe.opt_vqe(h_dvr_pauli, ansatzes[i], vqe_options)
        print(best_params)
        if (len(best_energy) < n_select) or (np.any(best_energy > np.min(best_energies))):
            best_gates.append(i)
            best_energy.append(np.min(best_energies))
            if len(best_energy) > n_select:
                del best_energy[np.argmax(best_energy)]
                del best_gates[np.argmax(best_energy)]
        print(best_energy)
        print('-----------------------')
    
    print(best_gates)
    for g in best_gates:
        print(gate_names[g], gate_list[g])

    ansatz0s_new = []
    touched_qubits_new = [touched_qubits[i].copy() for i in range(n_select)]
    for a, best_gate in enumerate(best_gates):
        touched_qubits_new[a] = touched_qubits[gate_names[best_gate][0]].copy()
        ansatz0 = ansatz0s[gate_names[best_gate][0]].copy()
        # print(a, touched_qubits_new)
        if gate_names[best_gate][1] == 'cnot':
            ansatz0.cnot(*gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate][0])
            touched_qubits_new[a].add(gate_list[best_gate][1])
            
        if gate_names[best_gate][1] == 'cz':
            ansatz0.cz(*gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate][0])
            touched_qubits_new[a].add(gate_list[best_gate][1])

        elif gate_names[best_gate][1] == 'ry':
            theta = Parameter(f'$\\theta$')
            ansatz0.assign_parameters({ansatz0.parameters[0]: theta}, inplace=True)
            ansatz0.ry(gate_names[best_gate][2] * theta, gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

        elif gate_names[best_gate][1] == '-ry':
            theta = Parameter(f'$\\theta$')
            ansatz0.assign_parameters({ansatz0.parameters[0]: theta}, inplace=True)
            ansatz0.ry(-theta, gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

        elif gate_names[best_gate][1] == 'h':
            ansatz0.h(gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

        elif gate_names[best_gate][1] == 'y':
            ansatz0.y(gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

        elif gate_names[best_gate][1] == 'x':
            ansatz0.x(gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

        elif gate_names[best_gate][1] == 'rz':
            theta = Parameter(f'$\\theta$')
            ansatz0.assign_parameters({ansatz0.parameters[0]: theta}, inplace=True)
            ansatz0.rz(theta, gate_list[best_gate])
            ansatz0s_new.append(ansatz0)
            touched_qubits_new[a].add(gate_list[best_gate])

    ansatz0s = ansatz0s_new
    touched_qubits = touched_qubits_new
    layer_energies.append(np.min(best_energy))
    
    for i in range(len(ansatz0s)):
        print(ansatz0s[i])
        print(touched_qubits[i])
        print(best_energy[i])
    print('--------------------------------------------')
```
